```python
class Chocolate:
    def __init__ (self, price: float, nachinka: str = None):
        if price <=0:
            raise ValueError ("Цена баточника должна быть больше нуля")
        self.price = price
        self.nachinka = nachinka

    def show_my_chocolate(self):
        if self.nachinka is None:
            print ("Обычный шоколадный батончик")
        else:
            print (f"Шоколадный батончик, добавка: {self.nachinka}")

    def __str__(self):
        return f"Шоколадный батончик, {self.price} рублей"

if __name__ == "__main__":
    # Батончик без добавки
    ch1 = Chocolate(59.9)
    ch1.show_my_chocolate()
    print(ch1)

    print()

    ch2 = Chocolate(89.9, "орехи")
    ch2.show_my_chocolate()
    print(ch2)

    print()

    ch3 = Chocolate(125.0, "карамель")
    ch3.show_my_chocolate()
    print(ch3)

    print()

    #Отработка исключения, если цена <=0
    try:
        ch_ne_ta_cena = Chocolate (-5.0, "ягоды")
    except ValueError as e:
        print (f"Ошибка: {e}")
```
